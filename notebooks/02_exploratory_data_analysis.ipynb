# Exploratory Data Analysis (EDA)

## Business Objective

The objective of this notebook is to explore the dataset and identify business patterns that can improve pricing, inventory planning, and demand forecasting.

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

df = pd.read_csv("../data/raw/retail_store_inventory (1).csv")

df.head()
```

```python
category_sales = (
    df.groupby("Category")["Units Sold"]
      .sum()
      .sort_values(ascending=False)
)

category_sales
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Units Sold by Category")
plt.xlabel("Category")
plt.ylabel("Units Sold")

plt.show()
```

## Business Insight

This analysis identifies the highest-selling product categories.

Business managers can prioritize inventory allocation, promotions, and pricing strategies for these categories.

```python
region_sales = (
    df.groupby("Region")["Units Sold"]
      .sum()
      .sort_values(ascending=False)
)

region_sales
```

```python
plt.figure(figsize=(8,5))

region_sales.plot(kind="bar")

plt.title("Units Sold by Region")
plt.show()
```

```python
avg_price = (
    df.groupby("Category")["Price"]
      .mean()
      .sort_values(ascending=False)
)

avg_price
```

```python
plt.figure(figsize=(8,5))

avg_price.plot(kind="bar")

plt.title("Average Product Price")

plt.show()
```
