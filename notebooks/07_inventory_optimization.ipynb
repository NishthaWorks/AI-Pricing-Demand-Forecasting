# Inventory Optimization Engine

## Objective

This notebook identifies inventory risks and provides AI-driven inventory recommendations to reduce stockouts, avoid overstocking, and improve inventory efficiency.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/retail_store_inventory (1).csv")
```

```python
def stock_status(row):

    if row["Inventory Level"] < row["Demand Forecast"]:
        return "Stockout Risk"

    elif row["Inventory Level"] > row["Demand Forecast"] * 2:
        return "Overstock"

    else:
        return "Healthy"

df["Stock Status"] = df.apply(stock_status, axis=1)

df["Stock Status"].value_counts()
```

```python
df["Recommended Reorder"] = np.where(
    df["Inventory Level"] < df["Demand Forecast"],
    np.ceil(df["Demand Forecast"] - df["Inventory Level"]),
    0
)

df[[
    "Inventory Level",
    "Demand Forecast",
    "Recommended Reorder"
]].head()
```

```python
def inventory_action(row):

    if row["Stock Status"] == "Stockout Risk":
        return "Order Immediately"

    elif row["Stock Status"] == "Overstock":
        return "Pause Purchasing"

    else:
        return "Maintain Inventory"

df["Inventory Recommendation"] = df.apply(
    inventory_action,
    axis=1
)

df["Inventory Recommendation"].value_counts()
```
