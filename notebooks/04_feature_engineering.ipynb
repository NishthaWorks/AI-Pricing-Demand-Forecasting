# Feature Engineering

## Objective

The objective of this notebook is to transform raw retail data into business-driven features that improve AI model performance.

Rather than using only existing columns, we will create intelligent business features that represent pricing strategy, inventory health, and market competitiveness.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/retail_store_inventory (1).csv")

# Revenue
df["Revenue"] = df["Units Sold"] * df["Price"]

df.head()
```

```python
df["Price Difference"] = df["Price"] - df["Competitor Pricing"]

df[["Price", "Competitor Pricing", "Price Difference"]].head()
```

```python
df["Inventory Coverage"] = (
    df["Inventory Level"] /
    (df["Units Sold"] + 1)
)

df["Inventory Coverage"].head()
```

```python
df["Forecast Error"] = (
    df["Demand Forecast"] -
    df["Units Sold"]
)

df["Forecast Error"].head()
```

```python
conditions = [
    df["Inventory Level"] < df["Units Sold"],
    df["Inventory Level"] > (df["Units Sold"] * 2)
]

choices = [
    "Stockout Risk",
    "Overstock Risk"
]

df["Inventory Risk"] = np.select(
    conditions,
    choices,
    default="Balanced"
)

df["Inventory Risk"].value_counts()
```

```python
df["Discount Percentage"] = (
    df["Discount"] / df["Price"]
) * 100

df["Discount Percentage"].head()
```
